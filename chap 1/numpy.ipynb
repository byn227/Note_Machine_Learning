```python
u=np.array([2,4,5,6])
v=np.array([1,0,3,4])
```

```python
def vector_multiplication(u,v):
    assert u.shape[0]==v.shape[0]
    n=u.shape[0]
    result=0.0
    for i in range(n):
        result=result+u[i]*v[i]
    return result
```

```python
vector_multiplication(u,v)
U=np.array([[1,2,3,5],[4,5,6,8],[7,8,9,10]])
```

```python
def matrix_vector_multiplication(m,v):
    assert m.shape[1]==v.shape[0]
    n_row=m.shape[0]
    result=np.zeros(n_row)
    for i in range(n_row):
            result[i]=vector_multiplication(m[i],v)
    return result
```

```python
matrix_vector_multiplication(U,v)
```

```python
U.dot(v)
V=np.array([[1,1,2],[0,1,0],[3,0,1],[4,2,1]])
```

```python
def matrix_matrix_multiplication(U,V):
    assert U.shape[1]==V.shape[0]
    n_row=U.shape[0]
    n_col=V.shape[1]
    result=np.zeros((n_row,n_col))
    for i in range(n_row):
        vi=V[:,i]
        result[:,i]=matrix_vector_multiplication(U,vi)
    return result
```

```python
matrix_matrix_multiplication(U,V)
```
