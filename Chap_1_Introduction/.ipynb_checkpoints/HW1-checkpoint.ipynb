```python
import numpy as np
```

```python
df_asia = df[df['origin'] == 'Asia']
```

```python
df_selected = df_asia[['vehicle_weight', 'model_year']]
```

```python
df_7 = df_selected.head(7)
```

```python
X = df_7.to_numpy()
```

```python
print(X)
```

```python
XTX = X.T @ X
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
```

```python
print(w.sum())
```
